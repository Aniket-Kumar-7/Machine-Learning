```python
df=pd.read_csv(r'/content/train.txt',sep=";",names=['text','emotion'],header=0)
```

```python
df.isnull().sum()
```

```python
emotion_list=df['emotion'].unique()
print(type(emotion_list))
```

```python
df['text']=df['text'].apply(lambda x:x.lower())
df['emotion']=df['emotion'].apply(lambda x:x.lower())
```

```python
import string
```

```python
def remove_punc(text):
  return text.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
df['emotion']=df['emotion'].apply(remove_punc)
```

```python
emotion_number={}
i=0
for emotion in emotion_list:
  emotion_number[emotion]=i
  i=i+1

df['emotion']=df['emotion'].map(emotion_number)
```

```python
emotion_number
```

```python
def remove_num(txt):
  new_txt=""
  for i in txt:
    if not i.isdigit():
      new_txt=new_txt+i
  return new_txt
```

```python
df['text']=df['text'].apply(remove_num)
```

```python
import re
```

```python
def advanced_prepr(text):
  text=text.lower()
  text=re.sub(r'https?://\S+|www\.\S+','',text);
  text=re.sub(r'<.*?>','',text)
  text=re.sub(r'@\w+|#\w+','',text)
  text=re.sub(r'\d+','',text)
  text=text.translate(str.maketrans('','',string.punctuation))
  text=re.sub(r'\s+',' ',text).strip()
  return text
```

```python
df['text']=df['text'].apply(advanced_prepr)
```

```python
df.head()
```

```python

def remove_emojis(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"
        "\U0001F300-\U0001F5FF"  # symbols 🌍
        "\U0001F680-\U0001F6FF"  # transport 🚗
        "\U0001F700-\U0001F77F"  # alchemical
        "\U0001F780-\U0001F7FF"
        "\U0001F800-\U0001F8FF"
        "\U0001F900-\U0001F9FF"  # supplemental 🤖
        "\U0001FA00-\U0001FA6F"
        "\U0001FA70-\U0001FAFF"
        "\U00002700-\U000027BF"  # dingbats ✂️
        "\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub(r'', text)
```

```python
df['text']=df['text'].apply(remove_emojis)
```

```python
from nltk.corpus import stopwords
```

```python
from nltk.tokenize import word_tokenize
import nltk
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
stop_words=set(stopwords.words('english'))
```

```python
df.head()
```

```python
contractions = {
    "im": "i am",
    "i m": "i am",
    "dont": "do not",
    "don t": "do not",
    "cant": "cannot",
    "can t": "cannot",
    "idk": "i do not know",
    "u": "you",
    "r": "are",
    'ive':"i have",
    'I':'i',
    'wasnt':'was not',
    'werent':'were not'
}
```

```python
def remove_slang(text):
  for key, value in contractions.items():
    text = re.sub(rf'\b{key}\b', value, text)
  return text
```

```python
df['text']=df['text'].apply(remove_slang)
```

```python
stop_words
```

```python
def remove_stopwords(text):
  words=text.split()
  cleaned=[]
  for i in words:
    if not i in stop_words:
      cleaned.append(i)
  return ' '.join(cleaned)
```

```python
print(remove_stopwords(df.loc[1]['text']))
```

```python
df['text']=df['text'].apply(remove_stopwords)
```

```python
print(df)
```

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
from sklearn.model_selection import train_test_split
```

```python
X=df['text']
y=df['emotion']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.20,random_state=42)
```

```python
y_train
```

```python
bow_vectorizer=CountVectorizer()
```

```python
X_train_bow=bow_vectorizer.fit_transform(X_train)
X_test_bow=bow_vectorizer.transform(X_test)
```

```python
X_test_bow
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

multi_nb_model=MultinomialNB()

multi_nb_model.fit(X_train_bow,y_train)

nb_pred=multi_nb_model.predict(X_test_bow)
```

```python
print(accuracy_score(nb_pred,y_test))
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lg_model=LogisticRegression(solver='lbfgs',max_iter=1000)
```

```python
lg_model.fit(X_train_bow,y_train)
```

```python
lg_pred=lg_model.predict(X_test_bow)
```

```python
print(accuracy_score(lg_pred,y_test))
```
