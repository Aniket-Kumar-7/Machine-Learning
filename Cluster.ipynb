```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sbn
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
```

```python
X,y=make_blobs(n_samples=500,n_features=2,centers=3,random_state=42,cluster_std=1.5)
```

```python
df=pd.DataFrame(X,columns=['column1','column2'])
```

```python
df
```

```python
scale_std=StandardScaler()
```

```python
X_scale=scale_std.fit_transform(X)
```

```python
X_scale
```

```python
ctr_list=[] #inertia
```

```python
K_range=np.arange(1,11)
```

```python
K_range
```

```python
for k in K_range:
  kmeans=KMeans(n_clusters=k,random_state=42)
  kmeans.fit(X_scale)
  ctr_list.append(kmeans.inertia_)
```

```python
ctr_list
```

```python
plt.plot(K_range,ctr_list,marker='o')
plt.show()
```

```python
# plt.plot(K_range,lbl_list,marker='o')
```

```python
kmeans_final=KMeans(n_clusters=3,random_state=42)
```

```python
cluster_labels=kmeans_final.fit_predict(X_scale)
```

```python
df['clusters']=cluster_labels
```

```python
sbn.scatterplot(x=df['column1'],y=df['column2'],hue=df['clusters'],palette='Set1')
```

```python
from sklearn.datasets import make_moons
```

```python
X,y=make_moons(n_samples=500,noise=.05,random_state=42)
```

```python
df=pd.DataFrame(X,columns=['column1','column2'])
```

```python
X_scale=scale_std.fit_transform(df)
```

```python
from sklearn.cluster import DBSCAN
```

```python
dfscan=DBSCAN(eps=.5,min_samples=5)
```

```python
cluster_labels=dfscan.fit_predict(X_scale)
```

```python
df['cluster_labels']=cluster_labels
```

```python
sbn.scatterplot(x=df['column1'],y=df['column2'],hue=df['cluster_labels'],palette='Set1')
```
