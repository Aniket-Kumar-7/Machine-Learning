```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sbn
import warnings
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score,r2_score,classification_report,confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
warnings.filterwarnings('ignore')
```

```python
data=pd.read_csv(r'/content/heart_attack_prediction_india.csv')
```

```python
data.shape
```

```python
data.info()
```

```python
cols=["Patient_ID", "State_Name", "Emergency_Response_Time", "Heart_Attack_History"]
```

```python
data.duplicated().sum()
```

```python
data.drop(["Patient_ID", "State_Name", "Emergency_Response_Time", "Heart_Attack_History"],axis=1,inplace=True)
```

```python
data.head()
```

```python
data.drop(['Annual_Income','Health_Insurance','Healthcare_Access','Air_Pollution_Exposure'],axis=1,inplace=True)
```

```python
data.head()
```

```python
data['Gender']=data['Gender'].map({'Male':1,'Female':0})
```

```python
X=data.drop('Heart_Attack_Risk',axis=1)
y=data['Heart_Attack_Risk']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.20,random_state=42)
```

```python
scaler=StandardScaler()
```

```python
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
liner_model=LinearRegression()
```

```python
liner_model.fit(X_train,y_train)
```

```python
y_pred_LinReg=liner_model.predict(X_test)
```

```python
y_pred_LinReg
```

```python
logistic_model=LogisticRegression()
```

```python
logistic_model.fit(X_train_scaled,y_train)
```

```python
y_pred_logstc=logistic_model.predict(X_test_scaled)
```

```python
y_pred_logstc
```

```python
confusion_matrix(y_pred_logstc,y_test)
```

```python
print(classification_report(y_pred_logstc,y_test))
```

```python
model_knn=KNeighborsClassifier(n_neighbors=5)
```

```python
model_knn.fit(X_train_scaled,y_train)
```

```python
y_pred_knn=model_knn.predict(X_test_scaled)
```

```python
print(classification_report(y_pred_knn,y_test))
```

```python
model_nb=GaussianNB()
```

```python
model_nb.fit(X_train_scaled,y_train)
```

```python
y_pred_nb=model_nb.predict(X_test_scaled)
```

```python
print(classification_report(y_pred_nb,y_test))
```

```python
model_svc=SVC(kernel='rbf')
```

```python
model_svc.fit(X_train_scaled,y_train)
```

```python
y_pred_svm=model_svc.predict(X_test_scaled)
```

```python
print(classification_report(y_pred_svm,y_test))
```

```python
model_dt=DecisionTreeClassifier()
```

```python
model_dt.fit(X_train,y_train)
```

```python
y_pred_dt=model_dt.predict(X_test)
```

```python
print(classification_report(y_pred_dt,y_test))
```
